```python

# ==============================
# DATA PREPARATION
# ==============================

# 1. City and Company Size counts
city_size_counts = pd.crosstab(
    data["City"],
    data["Size_of_Company"]
)

# 2. Funding percentage by Company Size
funding_size_percentage = pd.crosstab(
    data["Size_of_Company"],
    data["Funding_Type"],
    normalize="index"
) * 100

# 3. Top 8 categories
top_8_categories = (
    data["Category"]
    .value_counts()
    .head(8)
)

# 4. Funding profile
funding_profile = (
    data["Funding_Type"]
    .value_counts()
    .head(7)
)


# ==============================
# 3 × 2 CHART LAYOUT
# ==============================

fig, axes = plt.subplots(
    3, 2,
    figsize=(20, 20)
)


# ------------------------------------------------
# 1. BUBBLE CHART
# ------------------------------------------------

bubble_data = city_size_counts.reset_index().melt(
    id_vars="City",
    var_name="Company_Size",
    value_name="Company_Count",
)
bubble_colors=["#53263B", "#006400", "#800001", "#4b0082"]
city_names = bubble_data["City"].unique()
size_names = bubble_data["Company_Size"].unique()

city_position = {
    city: position
    for position, city in enumerate(city_names)
}

size_position = {
    size: position
    for position, size in enumerate(size_names)
}

bubble_x = bubble_data["City"].map(city_position)
bubble_y = bubble_data["Company_Size"].map(size_position)

bubble_colors = {
    "Small": "steelblue",
    "Medium": "lightcoral",
    "Startup": "mediumpurple",
    "Large": "darkseagreen"
}

for company_size in size_names:

    selected_data = bubble_data[
        bubble_data["Company_Size"] == company_size
    ]

    axes[0, 0].scatter(
        selected_data["City"].map(city_position),
        selected_data["Company_Size"].map(size_position),
        s=selected_data["Company_Count"] * 1.5 + 30,
        color=bubble_colors.get(company_size, "slategray"),
        alpha=0.65,
        label=company_size
    )

axes[0, 0].set_title("Company Size Concentration by City")
axes[0, 0].set_xlabel("City")
axes[0, 0].set_ylabel("Company Size")

axes[0, 0].set_xticks(range(len(city_names)))
axes[0, 0].set_xticklabels(
    city_names,
    rotation=45,
    ha="right"
)

axes[0, 0].set_yticks(range(len(size_names)))
axes[0, 0].set_yticklabels(size_names)

axes[0, 0].legend(title="Company Size")


# ------------------------------------------------
# 2. 100% STACKED BAR
# ------------------------------------------------

funding_size_percentage.plot(
    kind="bar",
    stacked=True,
    ax=axes[0, 1],
    color=[
        "#264653",
        "#8b0000",
        "#006400",
        "#4b0082",
        "#8b4513",
        "#791f49",
        "#6b362f"
    ]
)

axes[0, 1].set_title("Funding Mix by Company Size")
axes[0, 1].set_xlabel("Company Size")
axes[0, 1].set_ylabel("Percentage (%)")
axes[0, 1].set_ylim(0, 100)

axes[0, 1].tick_params(axis="x", rotation=0)

axes[0, 1].legend(
    title="Funding Type",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)


# ------------------------------------------------
# 3. KDE PLOT
# ------------------------------------------------

company_size_order = [
    "Startup",
    "Small",
    "Medium",
    "Large"
]

kde_colors = [
    "#0a2142",
    "#A02D2D",
    "#355E3B",
    "#3E0B53"
]

for company_size, chart_color in zip(
    company_size_order,
    kde_colors
):

    company_years = data.loc[
        data["Size_of_Company"] == company_size,
        "Year_Founded"
    ]

    sns.kdeplot(
        company_years,
        ax=axes[1, 0],
        color=chart_color,
        fill=True,
        alpha=0.25,
        label=company_size
    )

axes[1, 0].set_title("Founding Year Density by Company Size")
axes[1, 0].set_xlabel("Year Founded")
axes[1, 0].set_ylabel("Density")
axes[1, 0].legend(title="Company Size")


# ------------------------------------------------
# 4. SWARM PLOT
# ------------------------------------------------

swarm_sample = data.sample(
    n=min(800, len(data)),
    random_state=42
)

sns.swarmplot(
    data=swarm_sample,
    x="Software_Type",
    y="Year_Founded",
    hue="Software_Type",
    palette=[
        "#1F35F5",
        "#E95A18",
        "#0CB425",
        "#AD05F5"
    ],
    legend=False,
    size=3,
    ax=axes[1, 1]
)

axes[1, 1].set_title("Founding Years by Software Type")
axes[1, 1].set_xlabel("Software Type")
axes[1, 1].set_ylabel("Year Founded")

axes[1, 1].tick_params(
    axis="x",
    rotation=25
)


# ------------------------------------------------
# 5. RADAR CHART
# ------------------------------------------------

radar_labels = funding_profile.index.tolist()
radar_values = funding_profile.values.tolist()

radar_values = radar_values + radar_values[:1]

angles = np.linspace(
    0,
    2 * np.pi,
    len(radar_labels),
    endpoint=False
).tolist()

angles = angles + angles[:1]

radar_axis = fig.add_subplot(
    3, 2, 5,
    polar=True
)

radar_axis.plot(
    angles,
    radar_values,
    linewidth=2,
    color="#14F114"
)

radar_axis.fill(
    angles,
    radar_values,
    color="skyblue",
    alpha=0.35
)

radar_axis.set_xticks(
    angles[:-1]
)

radar_axis.set_xticklabels(
    radar_labels,
    fontsize=8
)

radar_axis.set_title(
    "Funding Type Profile",
    pad=20
)


# ------------------------------------------------
# 6. POLAR BAR CHART
# ------------------------------------------------

category_labels = top_8_categories.index.tolist()
category_values = top_8_categories.values

polar_axis = fig.add_subplot(
    3, 2, 6,
    polar=True
)

category_angles = np.linspace(
    0,
    2 * np.pi,
    len(category_labels),
    endpoint=False
)

polar_axis.bar(
    category_angles,
    category_values,
    width=0.55,
    color=[
    "#264653",  # Dark Blue
    "#8B0000",  # Dark Red
    "#4B0082",  # Dark Purple
    "#006400",  # Dark Green
    "#8B4513",  # Dark Brown
    "#C2185B",  # Dark Pink
    "#005F73",  # Dark Teal
    "#483D8B"   # Dark Slate Blue

    ],
    alpha=0.85
)

polar_axis.set_xticks(category_angles)

polar_axis.set_xticklabels(
    category_labels,
    fontsize=8
)

polar_axis.set_title(
    "Top 8 Software Categories",
    pad=20
)


# ==============================
# MAIN TITLE
# ==============================

plt.suptitle(
    "SaaS & Software Company Analytics — Advanced Insights",
    fontsize=20,
    fontweight="bold"
)

plt.tight_layout(
    rect=[0, 0, 1, 0.96]
)

plt.show()
```
